```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
from pathlib import Path
import sys

pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")

print(f"Python: {sys.executable}")
```

```python
# Toronto Open Data CKAN API — no auth needed
base_url = "https://ckan0.cf.opendata.inter.prod-toronto.ca"
package_id = "996cfe8d-fb35-40ce-b569-698d51fc683b"  # TTC Subway Delay Data

response = requests.get(
    f"{base_url}/api/3/action/package_show",
    params={"id": package_id}
)
package = response.json()["result"]

print(f"Dataset: {package['title']}")
print(f"\nResources available:")
for r in package["resources"]:
    print(f"  - {r['name']} ({r['format']}) — datastore_active: {r.get('datastore_active', False)}")
```

```python
# Download the files we need


# The years we want: 2022, 2023, 2024, and 2025-so-far
# Plus the code descriptions reference

# 1. Make a set (not a list — we're checking membership, sets are faster) 
wanted_resources = {
    "ttc-subway-delay-data-2022",
    "ttc-subway-delay-data-2023",
    "ttc-subway-delay-data-2024",
    "TTC Subway Delay Data since 2025.csv",  # grab CSV version for 2025
    "Code Descriptions.csv",                 # CSV version of delay codes
}

# 2. Make sure data/raw exists (mkdir with parents=True, exist_ok=True).
dest = Path("../data/raw")
dest.mkdir(parents=True, exist_ok=True)

# 3. Loop through package["resources"]:
#    - Skip if the name isn't in our wanted set
#    - Build a target filename (sanitize spaces if needed)
#    - If file already exists, skip
#    - Otherwise: requests.get(resource["url"]), write bytes to file
for resource in package["resources"]:
    if resource["name"] not in wanted_resources:
        continue

    name = resource["name"]
    fmt = resource["format"].lower()
    
    # Build a clean filename
    safe_name = name.replace(" ", "_").replace("/", "-")
    if not safe_name.lower().endswith(f".{fmt}"):
        safe_name = f"{safe_name}.{fmt}"
    target = dest / safe_name
    
    if target.exists():
        print(f"Already exists: {target.name}")
        continue

    print(f"Downloading: {name}...")
    r = requests.get(resource["url"])
    r.raise_for_status()
    target.write_bytes(r.content)
    print(f"  → {target.name} ({len(r.content) / 1024:.1f} KB)")
    
# 4. Print the contents of data/raw at the end so we can see what landed.
print("\nFiles in data/raw/:")
for f in sorted(dest.iterdir()):
    print(f"  {f.name} ({f.stat().st_size / 1024:.1f} KB)")
```

```python
# Quick schema check across all 4 years
years_files = {
    2022: "../data/raw/ttc-subway-delay-data-2022.xlsx",
    2023: "../data/raw/ttc-subway-delay-data-2023.xlsx",
    2024: "../data/raw/ttc-subway-delay-data-2024.xlsx",
    2025: "../data/raw/TTC_Subway_Delay_Data_since_2025.csv",
}

for year, fp in years_files.items():
    d = pd.read_excel(fp) if fp.endswith(".xlsx") else pd.read_csv(fp)
    print(f"{year}: {d.shape} — {d.columns.tolist()}")
```

```python
#Combine all 4 years into one DataFrame.

# 1. Start with an empty list called `dfs` (will hold each year's DataFrame).
dfs = []

# 2. Loop through years_files.items(). For each (year, filepath):
#    - Load with pd.read_excel if filepath ends in .xlsx, else pd.read_csv
#    - Drop the '_id' column if it exists:
#        d = d.drop(columns=['_id'], errors='ignore')
#      (2025 has it, others don't — errors='ignore' handles both cases)
#    - Add: d['source_year'] = year
#    - Append d to dfs
for year, fp in years_files.items():
    d = pd.read_excel(fp) if fp.endswith(".xlsx") else pd.read_csv(fp)
    d = d.drop(columns=['_id'], errors='ignore')
    d["source_year"] = year
    dfs.append(d)


# 3. Concatenate: df = pd.concat(dfs, ignore_index=True)
df = pd.concat(dfs, ignore_index=True)

# Normalize Date column — forces strings and timestamps into consistent datetime64

df['Date'] = pd.to_datetime(df['Date'])

# 4. Print:
#    - Combined shape
#    - Date range (df['Date'].min() and max())
#    - Rows per year: df['source_year'].value_counts().sort_index()
#    - Final column list
print(f"Combined shape: {df.shape}")
print(f"Date range: {df['Date'].min()} → {df['Date'].max()}")
print(f"\nRows per source year:")
print(df['source_year'].value_counts().sort_index())
print(f"\nColumns: {df.columns.tolist()}")
```

```python
df['datetime_str'] = df['Date'].dt.strftime('%Y-%m-%d') + ' ' + df['Time']
df['datetime'] = pd.to_datetime(df['datetime_str'], errors='coerce')
df = df.drop(columns=['datetime_str'])

n_bad = df['datetime'].isna().sum()
print(f"Rows with unparseable datetime: {n_bad}")
print(f"New datetime column dtype: {df['datetime'].dtype}")
print(f"\nSample:")
df[['Date', 'Time', 'datetime']].head(10)
```

```python
# Step 20: Combine all monthly weather CSVs into one DataFrame and clean it up.

# 1. Find all weather CSVs in data/raw/weather/ and sort them alphabetically.
#    Sorting matters because our filename format (YYYY_MM) sorts chronologically.
weather_files = sorted(weather_dir.glob("*.csv"))
print(f"Found {len(weather_files)} weather files")


# 2. Load each file and append to a list. We'll concat at the end.
#    Using a list-then-concat is much faster than growing a DataFrame in a loop.
weather_dfs = []
for fp in weather_files:
    w = pd.read_csv(fp)
    weather_dfs.append(w)


# 3. Concatenate into one big DataFrame.
weather = pd.concat(weather_dfs, ignore_index=True)
print(f"Combined shape: {weather.shape}")


# 4. Parse the Date/Time column into a proper datetime.
#    Rename it to 'datetime' so it's easier to work with (no spaces/parens).
weather['datetime'] = pd.to_datetime(weather['Date/Time (LST)'])


# 5. Drop columns we don't need to reduce noise:
#    - All the _Flag columns (data quality markers — not useful for our analysis)
#    - Precip. Amount (mm) — all NaN for this station (NAVCAN doesn't report it)
#    - Station metadata (Longitude, Latitude, Station Name, Climate ID) —
#      same value for every row, wasteful to keep
#    - Redundant time columns (we have datetime now)
cols_to_drop = [c for c in weather.columns if 'Flag' in c]
cols_to_drop += [
    'Precip. Amount (mm)',
    'Longitude (x)', 'Latitude (y)', 'Station Name', 'Climate ID',
    'Date/Time (LST)', 'Year', 'Month', 'Day', 'Time (LST)',
]
weather = weather.drop(columns=cols_to_drop, errors='ignore')


# 6. Rename the remaining columns to snake_case — easier to type and refer to.
weather = weather.rename(columns={
    'Temp (°C)': 'temp_c',
    'Dew Point Temp (°C)': 'dew_point_c',
    'Rel Hum (%)': 'humidity_pct',
    'Wind Dir (10s deg)': 'wind_dir_10deg',
    'Wind Spd (km/h)': 'wind_speed_kmh',
    'Visibility (km)': 'visibility_km',
    'Stn Press (kPa)': 'pressure_kpa',
    'Hmdx': 'humidex',
    'Wind Chill': 'wind_chill',
    'Weather': 'weather',
})


# 7. Sort by datetime for good measure. Useful if we ever plot time series.
weather = weather.sort_values('datetime').reset_index(drop=True)


# 8. Show the result.
print(f"\nFinal shape: {weather.shape}")
print(f"Date range: {weather['datetime'].min()} → {weather['datetime'].max()}")
print(f"\nColumns: {weather.columns.tolist()}")
weather.head(5)
```

```python
# Step 22: Join weather data onto the TTC delays.

# 1. Reload the combined delays from parquet (to make sure we have the latest).
df = pd.read_parquet("../data/processed/ttc_delays_combined.parquet")
print(f"Delays shape: {df.shape}")


# 2. Create a join key on delays: datetime rounded DOWN to the hour.
#    .dt.floor('h') takes a datetime and drops the minutes/seconds.
#    Example: 2024-01-01 02:08:00 → 2024-01-01 02:00:00
df['datetime_hour'] = df['datetime'].dt.floor('h')


# 3. The weather datetime is already on the hour, but make sure its dtype
#    matches what we just created. (Belt and suspenders.)
weather_for_join = weather.rename(columns={'datetime': 'datetime_hour'})


# 4. Left-join weather onto delays.
#    left join means: keep every delay row, add weather columns where available.
#    how='left' + on='datetime_hour' is the pandas equivalent of SQL:
#      SELECT * FROM df LEFT JOIN weather ON df.datetime_hour = weather.datetime_hour
merged = df.merge(weather_for_join, on='datetime_hour', how='left')
print(f"Merged shape: {merged.shape}")


# 5. Sanity check — how many delays couldn't find matching weather?
#    If weather covers the full date range, this should be 0 or very small.
unmatched = merged['temp_c'].isna().sum()
print(f"Delays with no weather match: {unmatched} ({unmatched/len(merged)*100:.2f}%)")


# 6. Quick peek at the joined data.
preview_cols = ['datetime', 'Station', 'Code', 'Min Delay', 'temp_c', 'weather']
merged[preview_cols].head(10)
```
